```python
import country_converter as coco
# Create a converter object
cc = coco.CountryConverter()

# Get a list of all countries in ISO3 format
countries = cc.data['ISO3'].dropna().unique().tolist()


# Create a dictionary to store countries and their regions
country_regions = {}

for country in countries:
    try:
        region = cc.convert(names=country, to='continent')
        country_name = cc.convert(names=country, to='name_short')
        country_regions[country_name] = region
    except ValueError:
        # Handle cases where region information is not available
        country_regions[country_name] = "Unknown"

# Print the country_regions dictionary
for country, region in country_regions.items():
    print(f"{country}: {region}")
```

```python
# Create a dictionary for country-region mapping
country_regions = {}

for country in countries:
    try:
        region = cc.convert(names=country, to='continent')
        country_name = cc.convert(names=country, to='name')
        country_regions[country_name] = region
    except ValueError:
        country_regions[country] = "Unknown"

# Print results
for country, region in country_regions.items():
    print(f"{country}: {region}")
```

```python
# Filter dataset to only include population-related indicators
df = df[df["Indicator Name"].str.contains("Population", case=False, na=False)]

# List of known non-country entries (regional, economic, or other groupings)
non_countries = [
    "Africa Western and Central", "Arab World", "Central Europe and the Baltics",
    "Channel Islands", "Caribbean small states", "East Asia & Pacific",
    "Early-demographic dividend", "Europe & Central Asia", "Euro area",
    "European Union", "Fragile and conflict affected situations", "High income",
    "Heavily indebted poor countries (HIPC)", "IBRD only", "IDA & IBRD total",
    "IDA total", "IDA blend", "IDA only", "Not classified",
    "Latin America & Caribbean", "Least developed countries: UN classification",
    "Low income", "Lower middle income", "Low & middle income", "Late-demographic dividend",
    "Middle East & North Africa", "Middle income", "North America", "OECD members",
    "Other small states", "Pre-demographic dividend", "Pacific island small states",
    "Post-demographic dividend", "South Asia", "Sub-Saharan Africa", "Small states",
    "World", "East Asia & Pacific (IDA & IBRD countries)", "Europe & Central Asia (IDA & IBRD countries)",
    "Latin America & the Caribbean (IDA & IBRD countries)", "Middle East & North Africa (IDA & IBRD countries)",
    "South Asia (IDA & IBRD)", "Sub-Saharan Africa (IDA & IBRD countries)", "Upper middle income"
]

# Remove rows that contain non-country regions
df = df[~df["Country Name"].isin(non_countries)]

# Function to convert only valid country names to continents
def get_continent(country):
    continent = cc.convert(names=country, to="continent")
    return continent if continent != "not found" else None  # Ignore invalid conversions

# Apply continent conversion
df["Continent"] = df["Country Name"].apply(get_continent)

# Remove rows where continent is missing
df = df.dropna(subset=["Continent"])

# Set the list of years to aggregate
years = [str(year) for year in range(1960, 2024)]  # Convert to string for column matching

# Group by Continent and sum population for each year
region_population = df.groupby("Continent")[years].sum().reset_index()

# Display result
print(region_population)
```

```python
#Bar plot for population distribution in the Continents for year 2023
continent_population_2023 = df.groupby('Continent')['2023'].sum().reset_index()

plt.bar(continent_population_2023['Continent'], continent_population_2023['2023'])
plt.xlabel('Continent')
plt.ylabel('Population in 2023')
plt.title('Population Distribution by Continent in 2023')
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels for better visibility
plt.show()
```

```python
# Convert country names to continents
df["Continent"] = df["Country Name"].apply(lambda x: cc.convert(names=x, to="continent"))

# Filter for each individual continent
africa_data = df[df["Continent"] == "Africa"]
america_data = df[df["Continent"] == "America"]
asia_data = df[df["Continent"] == "Asia"]
europe_data = df[df["Continent"] == "Europe"]
oceania_data = df[df["Continent"] == "Oceania"]

# Select relevant columns (years 1960-2023)
years = [str(year) for year in range(1960, 2024)]
africa_population = africa_data[years].sum().values  # Sum all countries in Africa
america_population = america_data[years].sum().values
asia_population = asia_data[years].sum().values
europe_population = europe_data[years].sum().values
oceania_population = oceania_data[years].sum().values
```

The populations of each continent have increased throughouth the years

```python
# top 10 most populous countries in 2023
top_countries = df[['Country Name', '2023']].nlargest(10, '2023')['Country Name'].values

# Filter dataset for only those countries
df_top = df[df['Country Name'].isin(top_countries)]
```

```python
# Convert wide format to long format
df_melted = df_top.melt(id_vars=['Country Name'], var_name='Year', value_name='Population')

# Keep only valid year columns (1960-2023)
df_melted = df_melted[df_melted['Year'].str.isnumeric()]
df_melted['Year'] = df_melted['Year'].astype(int)
```

```python
# Set up the figure
fig, ax = plt.subplots(figsize=(10, 6))
sns.set_style("whitegrid")

# Function to update the chart for each frame
def update(year):
    ax.clear()
    data = df_melted[df_melted['Year'] == year].sort_values(by='Population', ascending=True)
    ax.barh(data['Country Name'], data['Population'], color=sns.color_palette("husl", len(data)))
    ax.set_xlabel("Population")
    ax.set_title(f"Top 10 Most Populous Countries in {year}")
    ax.set_xlim(0, df_melted['Population'].max() * 1.1)

# Create animation
ani = animation.FuncAnimation(fig, update, frames=range(1960, 2024), repeat=True, interval=200)

# Save as GIF
ani.save("population_animation.gif", writer="pillow", fps=10)

# Show animation
plt.show()
```
